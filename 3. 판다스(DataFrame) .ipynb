## 변수 삭제(colimn 삭제 ): .drop()
- axis=0(default) - 행, 1 - 열
- inplace=False(default)

```python
# .drop : 복사본 (원본 xx)
# 열 통채로  지우려면  axis= 1
titanic_df.drop("Cabin", axis= 1)
# 여러개를 한꺼번에 지우려면  [] 로 감싸야 됌 
titanic_df.drop(["Cabin", "Embarked"], axis= 1)
```

```python
# 원본도  지우기 
titanic_df.drop(["Cabin", "Embarked"], axis= 1, inplace=True)
titanic_df
```

## 파생변수 만들기 

```python
# 확인 불가 
np.nan == np.nan
```

```python
# 만약 NaN이라면 :알수없음 
# 19세 이상: 성인 
# 19세 미만 : 아이 
# 조건식이 3개면 -> if~elif ~ else
def get_adult(age):
    if np.isnan(age):
        return "알 수 없음"   # 값이 없으면 "알 수 없음"으로 표시  
    return "성인" if age >= 19 else "아이"

get_adult(20)

```

```python
# .apply(): 앞의 시리즈의 값을 순차적으로 받아서 결과를 바꾸고 새로운 결과를 반환하는 함수
titanic_df["IsAdult"] = titanic_df["Age"].apply(get_adult)
titanic_df
```

```python
# RipOff
# Fare > 300 : 바가지 요금 
# Fare <= 300 : 바가지 요금 

# "RipOff"에 넣기만 함
titanic_df["RipOff"] = titanic_df["Fare"].apply(lambda fare: "바가지" if fare > 300 else "정상")

# (300을 넘는 승객들의 데이터만 )'필터링'해서 화면에 보여달라는 명령
titanic_df[titanic_df["Fare"] > 300]
```

```python
# 파생변수 IsAdult, RipOff 두 개를 모두 삭제 

titanic_df.drop(["IsAdult", "RipOff"], axis= 1)
# titanic_df.drop(["IsAdult", "RipOff"], axis= 1, inplace=True) # = 완전 삭제 
```
