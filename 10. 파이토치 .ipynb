## 내적(Dot Product) 계산 및 개념

### 개념
- 내적은 두 벡터 간의 원소별 곱을 모두 더하는 연산으로, 두 벡터가 얼마나 같은 방향을 향하는지 나타내는 유사도(방향성 일치도)의 척도입니다.

### 내적 공식
3차원 벡터 A = (x1, y1, z1)와 B = (x2, y2, z2)의 내적 공식은 다음과 같습니다.

A · B = x1*x2 + y1*y2 + z1*z3

- 첫 번째 성분끼리 곱하기: 1 x 11 = 11
- 두 번째 성분끼리 곱하기: 2 x 12 = 24
- 세 번째 성분끼리 곱하기: 3 x 13 = 39
- 최종 내적 값: 11 + 24 + 39 = 74 (0차원 스칼라)

### 내적과 행렬곱의 관계
1차원, 2차원, 3차원 텐서 모두 행렬곱 연산이 가능하며, 행렬곱의 결과 형태(Shape)는 입력되는 행렬들의 크기에 따라 다르게 결정됩니다. 행렬곱을 이루는 각 위치의 원소를 계산하는 본질적인 연산 방식이 바로 벡터의 내적이며, (1, n) 행렬과 (n, 1) 행렬의 곱처럼 결과가 1x1 형태의 2차원 행렬로 나오는 경우 그 내부 원소 값은 두 벡터의 내적 값(0차원 스칼라)과 동일합니다.

```python
정리!
행열곱은 행과 다른 행렬의 열을 곱해서 더하는 계산임
A = [1  2]      B = [5]
    [3  4]          [6]
A는 2행 2열 , B는 2행 1열인데  
```

```python
v1 = torch.tensor([1, 2, 3])
v2 = torch.tensor([1, 2, 3])

print(v1)
print(v2)
```

```python
print(torch.matmul(v1, v2))
```

```python
"""
    1차원 벡터 내적 [1, 2, 3] x [1, 2, 3]
    (3,) 크기를 가지는 1차원 벡터 2개의 내적 연산 과정입니다.

    - 첫 번째 성분끼리 곱하기: 1 x 1 = 1
    - 두 번째 성분끼리 곱하기: 2 x 2 = 4
    - 세 번째 성분끼리 곱하기: 3 x 3 = 9
    - 최종 연산 결과: 1 + 4 + 9 = 14
"""

None
```

```python
# 가로 백터 (1행 3열), 세로 백터 (3행 1열) 을  행렬곱해서 값 하나를 가진 1행 1열 행렬을 만듬 
# []의 갯수 = 행  \  안에 들어가 있는 숫자의 갯수 = 열

v3 = torch.tensor([[1, 2, 3]])
v4 = torch.tensor([[1], [2], [3]])

print(v3)
print(v4)


print(v3.shape)
print(v4.shape)

# [1 2 3] X [1]  = 1x1 + 2x2 + 3X3
# 1 + 4 + 9
# 14

print(torch.mm(v3, v4))
```

```python
# 크기가 같아서 덧셈, 행렬곱 둘다 가능
m3 = torch.tensor([[1, 2],[3, 4]])
m4 = torch.tensor([[5, 6],[7, 8]])

# [1 2][5 6] = [1X5 + 2X7 , 1X6 + 2X8]
# [3 4][7 8] = [3X5 + 4X6 , 3X6 + 4X8]

print(m3)
print(m4)
```

```python
# - 19 = [1, 2]와 [5, 7]
# - 22 = [1, 2]와 [6, 8]
# - 43 = [3, 4]와 [5, 7]
# - 50 = [3, 4]와 [6, 8]
print(torch.mm(m3, m4))
```

```python
"""
    2x2 행렬곱 [[1, 2], [3, 4]] x [[5, 6],
                                 [7, 8]]

    - 첫 번째 행 × 첫 번째 열:
      1 x 5 + 2 x 7 = 19

    - 첫 번째 행 × 두 번째 열:
      1 x 6 + 2 x 8 = 22

    - 두 번째 행 × 첫 번째 열:
      3 x 5 + 4 x 7 = 43

    - 두 번째 행 × 두 번째 열:
      3 x 6 + 4 x 8 = 50

    - 최종 연산 결과:
      [[19, 22], [43, 50]]

      내적이 클수록 → 같은 방향에 가까움
      내적이 작을수록 → 같은 반대에 가까움
"""

None
```

```python
a = torch.tensor([1, 2], dtype=torch.float32)
b = torch.tensor([3, 4], dtype=torch.float32)

diff = a - b
# [1, 2] - [3, 4]
# = [1-3, 2-4]
# = [-2, -2]
print(diff)

# torch.norm()은 벡터의 길이를 구함 
        # distance = √((-2)² + (-2)²)
        #          = √(4 + 4)
        #          = √8
        #          ≈ 2.8284
distance = torch.norm(diff)
print(distance)
```

## 파이토치의 핵심 장점: GPU 활용과 연산 가속
- GPU(Graphics Processing Unit)는 본래 이미지 렌더링과 같은 대규모 병렬 계산을 처리하기 위해 설계된 그래픽 처리 장치입니다.  
  하지만 최근에는 수많은 행렬 및 벡터 연산이 필요한 인공지능(AI)과 딥러닝 분야에서 뛰어난 연산 가속기로 널리 활용되고 있습니다.  
  GPU는 다수의 코어를 통해 수천, 수만 개의 연산을 동시에 병렬로 처리할 수 있는 구조를 갖추고 있어 딥러닝 모델 학습에 매우 최적화되어 있습니다.
- 파이토치에서 이러한 GPU 가속을 활용하기 위해서는 외장 그래픽카드와 함께 전용 하드웨어 설치 프로그램(예: NVIDIA CUDA 등)을 갖추어야 하며,  
  내장 그래픽 환경에서는 사용이 불가능합니다. 만약 로컬에 적합한 그래픽 카드가 없는 경우, 구글 코랩(Google Colab) 환경을 이용하면  
  무료로 GPU 가속을 사용해 볼 수 있습니다.
- 한편, 머신러닝과 딥러닝 전용으로 개발된 또 다른 가속 장치인 TPU(Tensor Processing Unit)의 경우,  
  본래 텐서플로(TensorFlow) 환경에 최적화되어 있어 파이토치에서는 직접적인 사용 및 적용이 다소 제한적입니다.  
  따라서 파이토치를 활용한 딥러닝 연산 시에는 GPU 환경을 구축하고 사용하는 것이 가장 일반적이고 효율적인 방법입니다.

```python
print("Cuda 사용 가능 여부:", torch.cuda.is_available())
print("Mps 사용 가능 여부:", torch.mps.is_available())
print("Cpu 사용 가능 여부:", torch.cpu.is_available())
print("Xpu(Itel 내장 그래픽)사용가능 여부 :", torch.xpu.is_available())
```

```python
if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.mps.is_available():
    device = torch.device("mps")
elif torch.xpu.is_available():
    device = torch.device("xpu")
else:
    device = torch.device("cpu")

print(f"현재 선택된 장치: {device}")
```

```python
t1 = torch.tensor([10])

# .to(device) : 텐서를 지정한 장치로 옮김 
t1_gpu = t1.to(device)

print(t1)      # -> CPU 메모리 
print(t1_gpu)  # -> GPU 메모리 

# cpu와 gpu의 영역이 다르기 때문에 연산이 안됨 
print(t1 + t1_gpu)
```

```python
print(t1.is_cpu)
print(t1_gpu.is_cuda)
print(t1_gpu.is_mps)
print(t1_gpu.is_xpu)
```

## 자동 미분과 기울기(requires_grad)
### 미분
- 어떤 값(변수)이 아주 조금 바뀔 때, 그에 따라 다른 값이 얼마나 변하는 지 계산하는 방법이다.

$$
\frac{\Delta y}{\Delta x}
$$

이는 x가 변한 만큼 y가 얼마나 변했는지를 나타냅니다.

예를 들어:

$$
x: 2 \rightarrow 3
$$

이면

$$
\Delta x = 1
$$

그리고

$$
y: 4 \rightarrow 7
$$

이면

$$
\Delta y = 3
$$

따라서

$$
\frac{\Delta y}{\Delta x}
=
\frac{3}{1}
=
3
$$

이것을 평균 변화율이라고 합니다.

미분은 변화량을 0에 가깝게 줄였을 때의 변화율을 의미합니다.

$$
\frac{dy}{dx}…

## 순전파 (Forward Pass): out 계산
### 1) 입력 벡터 정의
$$x = [x_1, x_2] = [3.0, 4.0]$$
$$y = [y_1, y_2] = [1.0, 2.0]$$

### 2) z = x + y 연산
각 성분별로 더해져 새로운 2차원 벡터 $z$가 생성됩니다.
$$z_1 = x_1 + y_1 = 3.0 + 1.0 = 4.0$$
$$z_2 = x_2 + y_2 = 4.0 + 2.0 = 6.0$$
$$z = [4.0, 6.0]$$

### 3) out = z.mean() 연산
벡터 $z$의 모든 성분을 더한 뒤 성분의 개수($n=2$)로 나누어 평균을 구합니다.
$$out = \frac{z_1 + z_2}{2} = \frac{4.0 + 6.0}{2} = 5.0$$

```python
# 순전파⭐️
# 각 가중치를 구해서 다 곱해서 out을 구하는 과정 
# 값과 가중치를 곱하고 곱한것끼리 더해서 y값을 구하는 것
```

## 2. 역전파 (Backward Pass): 미분 계산

`out.backward()`가 호출되면 연쇄 법칙(Chain Rule)에 의해 최종 출력($out$)에 대한 각 입력 성분들의 미분 값(기울기)이 역순으로 계산됩니다.

### 1) out에 대한 z의 편미분
$$out = \frac{1}{2}z_1 + \frac{1}{2}z_2$$
$$\frac{\partial out}{\partial z_1} = 0.5, \quad \frac{\partial out}{\partial z_2} = 0.5$$

### 2) 연쇄 법칙을 통한 최종 미분 계산 (x와 y 기준)
$z_1 = x_1 + y_1$ 이므로 $\frac{\partial z_1}{\partial x_1} = 1$ 입니다. 이를 활용해 최종 미분 값을 구합니다.

* **x에 대한 미분 값:**
  $$\frac{\partial out}{\partial x_1} = \frac{\partial out}{\partial z_1} \times \frac{\partial z_1}{\partial x_1} = 0.5 \times 1 = 0.5$$
  $$\frac{\partial out}{\partial x_2} = \frac{\partial out}{\partial z_2} \times \frac{\partial z_2}{\partial x_2} = 0.5 \times 1 = 0.5$$

* **y에 대한 미분 값 (위와 동일):**
  $$\frac{\partial out}{\partial y_1} = 0.5, \quad \frac{\partial out}{\partial y_2} = 0.5$$

```python
# x, y는 실수형 백터 ,
# equires_grad=True : 이 텐서가 결과에 얼마나 영향을 주는지 미분값을 계산(딥러닝 학습)
x = torch.tensor([3.0, 4.0], requires_grad=True) 
y = torch.tensor([1.0, 2.0], requires_grad=True) 

print(x, y)

# 순전파  
# (x + y) : 같은 위치끼리 더함 -> .mean : 평균 구함
# x + y = [3+1, 4+2] = [4, 6] -> (4+6) ÷ 2 = 5
# out = ((x1 + y1) + (x2 + y2)) ÷ 2
# out = (x1 + y1 + x2 + y2) ÷ 2

out = (x + y).mean()
print(out)
```

```python
# ⭐️ 역전파 
# out이 x, y에 대해 얼마나 변하는지 거꾸로 미분해서 계산

out.backward()
```

```python
# .grad는 “이 값이 결과에 얼마나 영향을 주는가”를 저장
print(f"x.grad: {x.grad}")
print(f"y.grad: {y.grad}")

# x의 첫 번째 값이 1 커지면 out은 0.5 커짐
# x의 두 번째 값이 1 커져도 out은 0.5 커짐
# y의 값들도 마찬가지
```
