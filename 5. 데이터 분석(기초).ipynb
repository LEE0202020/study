#### 1) 표본 조사 → 전체 조사
    - 과거에는 기술적 한계로 표본(필요한 것)만 조사 했지만, 기술이 발전하고 처리속도가 빨라져서 전체 조사할 수 있게 되었다.
    
#### 2) 사전처리보다 사후처리 중심 - 후
    - 과거에는 미리 정해진 규칙에 따라 쓸모없는 데이터를 버렸었지만
   데이터의 저장비용이 싸졌으므로 있는 그대로의 데이터를 전부 사용한다.
    
#### 3) 질(Quality)보다 양(Quantity) - 양
    - 데이터가 넘치고 전체가 보여주는 추세가 더 가치있다.

#### 4) 인과관계에서 상관관계를 분석 - 상
   - 비용이 비싸니 일을 시작하기 전에 인과 관계를 먼저 분석해야 했다.
   예를 들면 "사과 매출이 9월에 왜 증가했지?"를 먼저 분석해야 시작할 수 있었다.
   하지만 비용이 저렴해지면서 통계적으로 유의미한 상관관계를 찾아내고 비즈니스에 사용할 수 있게 되었다.

## 계급(Class)
- 변량을 일정 간격으로 나눈 구간을 의미한다
- 변량의 최솟값과 최댓값을 잘 고려해서 계급을 잘 정해야한다 

```python
df_seoul = df["서울"]

df_seoul
```

```python
df_class = pd.cut(df_seoul, bins=[140, 150, 160, 170, 180, 190])
df_class
```

```python
df_seoul_class = pd.DataFrame({"서울": df_seoul, "계급": df_class})
df_seoul_class
```

## 도수 
- 각 계급에 속하는 변량의 개수를 의미한다 

```python
df_seoul_class.groupby("계급").count()
```

```python
 # 각 계급에 속하는 변량의 비율
df_seoul_class.groupby("계급").count().apply(lambda x:x/100 )
```

## 도수 분포표 
- 주어진 자료를 계급별로 나눈 뒤 각 계급에 속하는 도수 및 상대도수를 조사한 표이다
- 구간별 분포를 한 번에 알아보기 좋지만 계급별 각 변량의 정확환 값이 생략되어 있다 

```python
frequency_df = pd.DataFrame({
    "도수": df_seoul_class.groupby("계급").count()["서울"],
    "상대 도수": df_seoul_class.groupby("계급").count().apply(lambda x: x / 100)["서울"]
})

frequency_df
```

## 편차 (Deviation)

```python
np.random.seed(30)

# 지역 별 100명의 키 데이터
df = pd.DataFrame(
    np.random.randint(140, 190, size=(10, 10)),
    columns="서울,경기,인천,광주,대구,부산,전주,강릉,울산,수원".split(",")
)
```

```python
gyeonggi_df = df["경기"].to_frame()
gyeonggi_df
```

```python
# 경기도 평균을 뺴기

gyeonggi_df["경기"].apply(lambda x: x - gyonggi_df.mean())
```

```python
gyeonggi_df["편차"] = gyeonggi_df["경기"].apply(lambda x: x - gyeonggi_df.mean())

gyeonggi_df
```

## 분산(VAriance)
- 이상치를 구하기 위해 평균으로부터 얼마나 떨어져 있는지 알아내기 위한 값

1. 절대값으로 구한다
2. 모든 편차의 제곱을 하고 그 합을 평균을 낸다 

```python
gyeonggi_df["편차의 제곱"] = gyeonggi_df["편차"].apply(lambda x: x ** 2)
```

```python
# 분산
variance = gyeonggi_df["편차의 제곱"].mean()     # .mean() :평균 
variance
# variance여기에서 루트를 씌운게 표준 편차임

# 제곱을 하는 이유는 음수를 양수로 하기 위함 
```

## 표준 편차(Standard deviation)
- 분산의 제곱근이며, 관측된 변량의 흩어진 정로를 하나의 수치로 나타내는 통계량이다.
- 표준편차가 작을 수록 평균값에서 변량들의 거리가 가깝다고 판단한다.

```python
import math 

# .sqrt: 스퀘어 루트 
std = math.sqrt(variance)
std
```
