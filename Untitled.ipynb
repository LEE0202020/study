```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "product": ["A", "B", "C", "D", "E", "F", "F"],
    "price": [12000, 15000, np.nan, 18000, 20000, 999999, 999999],
    "stock": [10, 5, 8, np.nan, 20, 3, 3],
    "category": ["식품", "전자", "식품", "전자", np.nan, "식품", "식품"],
    "sold": [5, 2, 7, 1, 9, 3, 3]
})
df
```

```python
# 1번. 결측치와 중복 데이터 전처리
# 1. 각 컬럼별 결측치 개수를 확인하시오.
# 2. price의 결측치를 price의 중앙값으로 대체하시오.
# 3. stock의 결측치를 stock의 평균값으로 대체하시오.
# 4. category의 결측치를 category의 최빈값으로 대체하시오.
# 5. 중복된 행을 제거하시오.
# 6. 인덱스를 0부터 다시 설정하시오.
```

```python
# 1. 각 컬럼별 결측치 개수를 확인하시오.
# df.fillna(df.mode())
# df.isna(df.mode())
df.isna().sum()
```

```python
# 2. price의 결측치를 price의 중앙값으로 대체하시오.
price_median = df["price"].median()
df["price"] = df["price"].fillna(price_median)
df
```

```python
# 3. stock의 결측치를 stock의 평균값으로 대체하시오.
stock_mean = df["stock"].mean()
df["stock"] = df["stock"].fillna(stock_mean)
df
```

```python
# 4. category의 결측치를 category의 최빈값으로 대체하시오.
category_mode = df["category"].mode()[0]
df["category"] = df["category"].fillna(category_mode)
df
```

```python
# 5. 중복된 행을 제거하시오.
df.drop_duplicates()
```

```python
# 6. 인덱스를 0부터 다시 설정하시오.
df.iloc()
```
