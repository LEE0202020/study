## Series
   - Pandas의 기본 객체 중 하나로, numpy의 ndarray를 기반으로 인덱싱 기능이 추가된
   1차원 배열을 나타낸다. (index x value의 형태)
   - index를 지정하지 않을 시, 기본적으로 ndarray와 같이 0-based 인덱스 생성한다
   - index를 지정할 경우 명시적으로 지정된 index를 사용한다.
   - 같은 타입의 0개 이상의 데이터를 가질 수 있다

[판다스 공식문서](https://pandas.pydata.org/)

```python
import numpy as np
import pandas as pd
```

```python
arr1 = np.array([1,2,3])
print(arr1)
print(type(arr1))
```

```python
pd.Series(arr1)   # Series : 클래스 -> 필드 가능  
```

```python
s1 = pd.Series(arr1, index= ['a', 'b', 'c'])   # 인덱스 자리에  a', 'b', 'c'들어감 , 이는 키값 임 
s1
```

```python
print(s1['a'])
```

```python
s2 = pd.Series([i * 10 for i in range(1, 6)])
s2
```

```python
# 넘파이로 변경함 
s2[0]
```
