## 2단계 , 데이터 분할 

```python
# 독립 변수 , 종속 변수 
X= fish_df.drop("Species", axis=1)
X.head()
```

```python
y = fish_df.drop("Species", axis=1)
X.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=2026)
print(X_train.shape, X_test.shape, y_test.shape, y_train.shape)
```

## 3단계 모델 준비

#  장점 
- 데이터만 있으면 알고리즘을 사용할 수 있다 

```python
from sklearn.model_selection import KNeighborsClassifier

kn = KNeighborsClassifier()
```
