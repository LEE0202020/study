## 브로드캐스팅
- 다른 크기의 배열들 간에도 자동으로 연산해주는 Numpy의 강력한 기능

## 브로드캐스팅의 특징
    - 기본적으로는 shape와 같은 두 ndarray에 대한 연산은 각 원소별로 진행한다
    - 다른 shape를 갖는 array 간의 연산의 경우 브로드 캐스팅(shape를 맞춤) 후 진행한다

<img width=500 src='https://www.astroml.org/_images/fig_broadcast_visual_1.png' />

```python
arr1 = np.arange(12).reshape(4, 3)
arr1

print(arr1.shape)
# (4행 3열)
```

```python
arr2 = np.arange(3)

print(arr2)
print(arr2.shape)
# 행 데이터 3개 
```

```python
print(arr1 + arr2)
```

## ⭐️ argmax()

```python
arr3 = np.arange(5, 10)
print(arr3)

# [5 6 7 8 9]에서 제일 큰 값은 9
print(arr3[arr3.argmax()])
```

```python
np.random.seed(918)
# np.raarr4.argmaxndom.randint(10, size=(0,6))
arr4 = np.random.randint(10, size=(1,6)).flatten()
print(arr4.argmax()) # 가장 큰 값의 인덱스
print(arr4.argmin())  # 가장 작은 값의 인덱스 
print(arr4.max())   # 가장 큰 값
print(arr4.min())   # 가장 작은 값
print(argmax(arr4))
```

```python
arr5 = np.random.randint(10, size=(3,6))
arr5
```

```python
# 출력 방향 x축 ,연산은 y축방향에 있는 숫자 연산
print(arr5.argmax(axis=0))

# 출력 방향 y축 ,연산은 x축방향에 있는 숫자 연산
print(arr5.argmax(axis=1))
```

## boolean selection, boolean indexing
- array 인덱싱 시 , bool 타입의 조건에 맞는 list를 전달하여 조건식을 만족하는 True 의 결과불만 필ㅌ링 한다 

```python
arr4 = np.random.randint(1, 100, size=20)
print(arr4)
```

```python
print(arr4 / 5)
print(arr4 % 2)  # 결가 가 0이 나오면 짝수 이고 1이 나오면 홀수

even_mask = arr4 % 2 == 0
print(even_mask)
```

## 다중 조건식 
- &: and 조건식
- |: or 조건식 


```python
# 홀수 이면서 50보다 작은 값
# and:
np.random.seed(2026)
arr5 = np.random.randint(100, size=15) 
arr5
```
